## 5. Make a plot and explain it

A plot helps you check what your calculations hide. This code draws the observations and their sample mean. The x-axis is a position in this toy sample; it is not a claim about a temperature trend.

````python
import numpy as np
import matplotlib.pyplot as plt
x = np.array([30, 32, 31, 35, 32])
plt.plot(np.arange(1, len(x) + 1), x, "o-", label="Daily high")
plt.axhline(x.mean(), color="tab:orange", label="Sample mean")
plt.xlabel("Observation")
plt.ylabel("Temperature (C)")
plt.legend()
plt.show()
````

## 7. Read a test result

A result needs a question and assumptions. Here the null is that the population mean is 30 C. Treat these synthetic observations as independent and approximately normally distributed for this exercise. Actual consecutive daily temperatures often violate independence; check persistence before using an ordinary t-test.

````python
import numpy as np
from scipy import stats
x = np.array([30, 32, 31, 35, 32], dtype=float)
result = stats.ttest_1samp(x, popmean=30)
print(f"t = {result.statistic:.3f}")
print("Reject H0 at 5%:", bool(result.pvalue < 0.05))
````

**Expected output**

```
t = 2.390
Reject H0 at 5%: False
```
